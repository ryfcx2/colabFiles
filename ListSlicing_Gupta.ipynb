#Part 1 - List Slicing


```python

pi = [3, 1, 4, 1, 5, 9]
e = [2, 7, 1]
```

0. Use pi and/or e to create the list [3, 7, 1],and assign it to the variable answer0.
1. Use pi and/or e to create the list [1, 5, 9] and assign it to the
variable answer1.
2. Use pi and/or e to create the list [3, 4, 5], and assign it to the variable answer2.
3. Use pi and/or e to create the list [5, 4, 3, 2, 1], and assign it to the variable answer3.
4. Use pi and/or e to create the list [9, 1, 1, 1], and assign it to the variable answer4.

```python
#problem 0:
answer0 = [pi[0]]+e[1:3:1]
print(answer0)

#problem 1:
answer1 = [e[2]]+pi[4:6:1]
print(answer1)

#problem 2:
answer2 = pi[0:6:2]
print(answer2)

#problem 3:
answer3 = pi[4:1:-2]+[pi[0]]+e[0:3:2]
print(answer3)

#problem 4:
answer4 = pi[5:0:-2]+[e[2]]
print(answer4)

```

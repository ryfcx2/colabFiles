```
Function 2: Write a function called contains_string(values, str) that accepts a
list of strings and another string as parameters and returns True if the list
contains the given string and false otherwise.
contains_string(['hello', 'cats', 'dogs'], 'cats') → True
contains_string(['hello', 'cats', 'dogs'], 'cat') → False
contains_string(['a', 'b', 'c'], 'd') → False
```

````python
def contains_string(values, str):
  ft=False
  for i in values:
    if i == str:
      ft = True
  return ft

print(contains_string(['hello', 'cats', 'dogs'], 'cats'))
print(contains_string(['hello', 'cats', 'dogs'], 'cat'))
print(contains_string(['a', 'b', 'c'], 'd'))
````

```
Function 3: Write a function called count_lengths(values, length) that accepts a
list of strings and an int as parameters and returns the number of strings in the
list that are exactly the given length.
count_lengths(['a', 'b', 'c'], 2) → 0
count_lengths(['a', 'aa', 'aaa'], 3) → 1
count_lengths(['cat', 'dog', 'bird'], 3) → 2
```

````python
def count_lengths(values, length):
  count =0
  for i in values:
    if len(i) == length:
      count+=1
  return count

print(count_lengths(['a', 'b', 'c'], 2))
print(count_lengths(['a', 'aa', 'aaa'], 3))
print(count_lengths(['cat', 'dog', 'bird'], 3))
````

Function 4: Write a function called match_up(list1, list2) that takes two lists as
parameters.
Given list1 and list2 of the same length, for every element in the first list,
consider the corresponding element in the other list(at the same index). Return the
count of the number of times that the two elements are not equal to each other.
match_up([1, 2, 3], [1, 2, 3]) → 0
match_up([1, 2, 3], [2, 3, 5]) → 3
match_up([1, 2, 3], [3, 3, 3]) → 2
'''```



````python
def match_up(list1, list2):
  count = 0
  for i in range(0, len(list1)):
    if list1[i] != list2[i]:
      count+=1
  return count

print(match_up([1, 2, 3], [1, 2, 3]))
print(match_up([1, 2, 3], [2, 3, 5]))
print(match_up([1, 2, 3], [3, 3, 3]))
````

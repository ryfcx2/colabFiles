Write a function every_other(s) that accepts a string s and returns a new string containing only the characters at even indices (0, 2, 4, ...)
every_other('computer') → 'cmue'
every_other('abcdefgh') → 'aceg'
every_other('hi') → 'h'


```python
def every_other(s):
  return s[::2]
print(every_other('computer'))
```

Write a function move_to_end(s, n) that accepts two parameters, string s and integer n, and that returns the a new string in which the first n characters of s have been moved to the end of the string. Hint: Think back to our list/string slicing practice with pi & e!
Examples:
move_to_end('computer', 3) → 'putercom'
move_to_end('computer', 5) → 'tercompu'
move_to_end('computer', 0) → 'computer'
Special case: If n (the second input) is greater than or equal to the length of s (the first input), then the function should simply return the original s without changing it:
move_to_end('hi', 5) → 'hi'

```python
def move_to_end(s, n):
  if n>=len(s):
    return s
  return s[n:]+s[:n]
print(move_to_end('computer', 3))
```

Write a function remove_middle(lst, start, end) that accepts a list lst and two integers start and end, and returns a new list with the elements from index start up to (but not including) index end removed.
Examples:
remove_middle([1, 2, 3, 4, 5, 6], 1, 3) → [1, 4, 5, 6]
remove_middle([1, 2, 3, 4, 5, 6], 0, 2) → [3, 4, 5, 6]
remove_middle(['a','b','c','d'], 2, 4) → ['a', 'b']

```python
def remove_middle(lst, start, end):
  return lst[:start] +lst[end:]
print(remove_middle([1, 2, 3, 4, 5, 6], 1, 3))
```

Write a function mirror(s) that accepts a string s as a parameter and returns a mirrored version of s that is twice the length of the original string. For example:
Examples:
mirror(‘bacon') → baconnocab
mirror(‘XYZ') → XYZZYX

```python
def mirror(s):
  nest = s+ s[::-1]
  return nest
```

Write a function is_palindrome_ignore_spaces(s) that accepts a string s and returns True if the string is a palindrome (reads the same forwards and backwards), ignoring spaces, and False otherwise. Assume the string is already lowercase.

Hint: You'll need a string method to deal with the spaces before you can use slicing to reverse it.

Examples:
is_palindrome_ignore_spaces('racecar') → True
is_palindrome_ignore_spaces('taco cat') → True
is_palindrome_ignore_spaces('python') → False

```python
def is_palindrome_ignore_spaces(s):
  s=s.replace(' ', '')
  if s == s[::-1]:
    return True
  else:
    return False
print(is_palindrome_ignore_spaces('taco cat'))
```

Pig Latin generator. Write a function pig_Latin(x) that accepts as a parameter a string, and returns the word in pig latin. It should be able to handle words that begin in consonants as well as words that begin in vowels. See below for examples of how to handle consonants vs vowels. TO MAKE IT EASIER
* You only need to write code to change one word to pig Latin
* Don’t worry if a word begins with a double consonant. In real pig Latin, “switch” should be “itch-sway”. For our generator, you can make “switch” into “witch say”
Examples:
pig_Latin(‘dog’) → og day
pig_Latin(‘ate’) → ate yay
pig_Latin(‘switch’) → witch say

```python
def pig_Latin(x):
  if x[0] in 'aeiou':
    return x +'yay'
  return x[1:]+" "+x[0]+'ay'
print(pig_Latin('dog'))
```
